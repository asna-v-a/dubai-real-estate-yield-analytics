```python
import pandas as pd

# Load both files
df1 = pd.read_csv("../data/raw/transactions_2026-08-21_18-30-31_0001.csv")
df2 = pd.read_csv("../data/raw/transactions_2026-08-21_18-30-31_0002.csv")

# Merge them into one
df = pd.concat([df1, df2], ignore_index=True)

print(df.shape)
print(df.columns.tolist())
df.head()
```

```python
# Check real district names
print(df['area_name_en'].nunique())
print(sorted(df['area_name_en'].dropna().unique())[:50])

# Check trans_group_en values (should mostly be "Sales" since that's what we filtered)
print(df['trans_group_en'].value_counts())

# Check property_usage_en values
print(df['property_usage_en'].value_counts())

# Check for missing values in key columns
print(df[['actual_worth', 'area_name_en', 'instance_date', 'procedure_area', 'meter_sale_price']].isnull().sum())
```

```python
# Search for each target district by keyword
targets_to_find = ["Downtown", "Marina", "Business Bay", "Jumeirah Village", "Palm Jumeirah", "Hills"]

for keyword in targets_to_find:
    matches = [name for name in df['area_name_en'].unique() if keyword.lower() in str(name).lower()]
    print(f"'{keyword}' matches: {matches}")
```

```python
targets_to_find_v2 = ["Burj Khalifa", "Marsa", "Thanyah", "Hadaeq", "Barsha South"]

for keyword in targets_to_find_v2:
    matches = [name for name in df['area_name_en'].unique() if keyword.lower() in str(name).lower()]
    print(f"'{keyword}' matches: {matches}")
```

```python
for area in ["Al Thanyah Fifth", "Al Thanyah Third", "Al Thanyah First"]:
    sample_buildings = df[df['area_name_en'] == area]['building_name_en'].dropna().unique()[:15]
    print(f"\n{area} sample buildings:")
    print(sample_buildings)
```

```python
jvc_building_keywords = ["Diamond Views", "Belgravia", "FIORE", "Green Park", "Bloom Towers"]

for keyword in jvc_building_keywords:
    matches = df[df['building_name_en'].str.contains(keyword, case=False, na=False)]
    if len(matches) > 0:
        print(f"'{keyword}' found in area(s): {matches['area_name_en'].unique()}")
    else:
        print(f"'{keyword}': no match")
```

```python
# Convert dates safely - invalid/corrupted dates become NaT (Not a Time) instead of crashing
df['instance_date_parsed'] = pd.to_datetime(df['instance_date'], errors='coerce')

# Check how many rows had bad dates
print(f"Rows with unparseable dates: {df['instance_date_parsed'].isna().sum()} out of {len(df)}")

target_districts = [
    "Burj Khalifa",
    "Marsa Dubai",
    "Business Bay",
    "Palm Jumeirah",
    "Hadaeq Sheikh Mohammed Bin Rashid",
    "Al Barsha South Fourth",
    "Al Barsha South Fifth"
]

df_filtered = df[
    (df['area_name_en'].isin(target_districts)) &
    (df['trans_group_en'] == 'Sales') &
    (df['property_usage_en'] == 'Residential') &
    (df['instance_date_parsed'] >= '2023-01-01')
].copy()

print(df_filtered.shape)
print(df_filtered['area_name_en'].value_counts())
```

```python
df_filtered.to_csv("../data/clean/transactions_filtered.csv", index=False)
print("Saved.")
```
